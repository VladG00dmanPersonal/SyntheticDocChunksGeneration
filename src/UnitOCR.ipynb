```python
import json

OCR_DIR = "../docs/Charter1/"
OCR_FILENAME = "ocr.json"
OUTPUT_FILENAME = "ocr.txt"
```

```python
with open(f"{OCR_DIR}{OCR_FILENAME}", "r") as file:
    ocr_json = json.load(file)

unit_content = [i["content"] for i in ocr_json]

with open(f"{OCR_DIR}{OUTPUT_FILENAME}", "w") as file:
    file.write("\n".join(unit_content))
```
